```python
from rdflib import Graph,URIRef,Namespace,BNode,Literal
from rdflib.namespace import XSD,RDF
import pandas as pd
import argparse
from collections import defaultdict
from datetime import datetime
```

```python
graph = Graph()
print("Started loading graph...")
graph.parse('sample.ttl', format="turtle")
print("Graph loaded successfully.")
```

```python
timeseries_unique_observations_id = []
timeseries_unique_observations_time = []

predicate = URIRef("http://www.w3.org/ns/sosa/resultTime")
```

```python
#for subj in graph.subjects(predicate=predicate):
    #timeseries_unique_observations_time.append(subj)
```

```python

for S,P,O in graph.triples((None, predicate, None)):
    #print(S)
    #print(O)
    timeseries_unique_observations_time.append(O)
    timeseries_unique_observations_id.append(S)


```

```python
mapped_values = dict(zip(timeseries_unique_observations_time,timeseries_unique_observations_id))
```

```python
# Parse into datetime objects
parsed_datetimes = [datetime.fromisoformat(dt) for dt in timeseries_unique_observations_time]

# Sort ascending
parsed_datetimes.sort()
```

```python
# Convert back to string if needed
sorted_xsd_strings = [dt.isoformat() for dt in parsed_datetimes]
```
